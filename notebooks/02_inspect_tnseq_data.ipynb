```python
ta_matches = []

for coordinate in table_s2["Coordinate"]:
    position = int(coordinate)

    dinucleotide = matching_sequence[position - 1 : position + 1]
    ta_matches.append(dinucleotide.upper() == "TA")

print("Coordinates tested:", len(ta_matches))
print("Coordinates matching TA:", sum(ta_matches))
print("Coordinates not matching TA:", len(ta_matches) - sum(ta_matches))
print(
    "Percentage matching TA:",
    round(100 * sum(ta_matches) / len(ta_matches), 4)
)
```

```python
coordinate_tests = {
    "Coordinate is T position (1-based)": lambda p: matching_sequence[p - 1:p + 1],
    "Coordinate is A position (1-based)": lambda p: matching_sequence[p - 2:p],
    "Coordinate is T position (0-based)": lambda p: matching_sequence[p:p + 2],
}

for test_name, get_dinucleotide in coordinate_tests.items():
    matches = 0

    for coordinate in table_s2["Coordinate"]:
        position = int(coordinate)
        dinucleotide = get_dinucleotide(position).upper()

        if dinucleotide == "TA":
            matches += 1

    percentage = 100 * matches / len(table_s2)

    print(test_name)
    print("Matches:", matches)
    print("Not matching:", len(table_s2) - matches)
    print("Percentage:", round(percentage, 4))
    print()
```

```python
# Inspect gene annotations from the available GFF file

gene_records = []

with matching_gff_file.open("r", encoding="utf-8") as handle:
    for line in handle:
        if line.startswith("#"):
            continue

        columns = line.rstrip("\n").split("\t")

        if len(columns) != 9:
            continue

        sequence_id, source, feature_type, start, end, score, strand, phase, attributes = columns

        if feature_type == "gene":
            gene_records.append(
                {
                    "sequence_id": sequence_id,
                    "start": int(start),
                    "end": int(end),
                    "strand": strand,
                    "attributes": attributes,
                }
            )

print("Number of genes in GFF:", len(gene_records))
gene_records[:3]
```

```python
def parse_gff_attributes(attribute_text):
    parsed = {}

    for item in attribute_text.split(";"):
        if "=" in item:
            key, value = item.split("=", 1)
            parsed[key] = value

    return parsed


parsed_attributes = genes_df["attributes"].apply(parse_gff_attributes)

genes_df["locus_tag"] = parsed_attributes.apply(
    lambda values: values.get("locus_tag")
)

genes_df["old_locus_tag"] = parsed_attributes.apply(
    lambda values: values.get("old_locus_tag")
)

genes_df["gene_name"] = parsed_attributes.apply(
    lambda values: values.get("gene")
)

genes_df[
    [
        "sequence_id",
        "start",
        "end",
        "strand",
        "locus_tag",
        "old_locus_tag",
        "gene_name",
    ]
].head(10)
```

```python
import re

def convert_rvbd_to_rv(old_locus_tag, strand):
    if old_locus_tag is None:
        return None

    match = re.fullmatch(r"RVBD_(\d{4})", str(old_locus_tag).strip())

    if match is None:
        return None

    number = match.group(1)
    suffix = "c" if strand == "-" else ""

    return f"Rv{number}{suffix}"


genes_with_rv["rv_id"] = genes_with_rv.apply(
    lambda row: convert_rvbd_to_rv(
        row["old_locus_tag"],
        row["strand"]
    ),
    axis=1
)

genes_with_rv[
    [
        "old_locus_tag",
        "rv_id",
        "start",
        "end",
        "strand",
        "gene_name",
    ]
].head(10)
```

```python
def get_gene_and_cds_records(orf_id, genes_with_rv, cds_df):
    """
    Find the gene and CDS records for a given Rv ORF ID.

    Parameters
    ----------
    orf_id : str
        ORF identifier, for example 'Rv0405' or 'Rv0020c'.
    genes_with_rv : pandas.DataFrame
        Gene annotation table containing rv_id, locus_tag, start, end and strand.
    cds_df : pandas.DataFrame
        CDS annotation table containing locus_tag and protein_id.

    Returns
    -------
    dict
        Dictionary containing gene and protein information.
    """

    gene_matches = genes_with_rv[
        genes_with_rv["rv_id"].eq(orf_id)
    ].copy()

    if len(gene_matches) == 0:
        return {
            "orf_id": orf_id,
            "status": "gene_not_found",
        }

    if len(gene_matches) > 1:
        return {
            "orf_id": orf_id,
            "status": "multiple_gene_matches",
            "number_of_matches": len(gene_matches),
        }

    gene_record = gene_matches.iloc[0]
    locus_tag = gene_record["locus_tag"]

    cds_matches = cds_df[
        cds_df["locus_tag"].eq(locus_tag)
    ].copy()

    if len(cds_matches) == 0:
        return {
            "orf_id": orf_id,
            "status": "cds_not_found",
            "locus_tag": locus_tag,
            "gene_start": gene_record["start"],
            "gene_end": gene_record["end"],
            "strand": gene_record["strand"],
        }

    if len(cds_matches) > 1:
        return {
            "orf_id": orf_id,
            "status": "multiple_cds_matches",
            "locus_tag": locus_tag,
            "number_of_matches": len(cds_matches),
        }

    cds_record = cds_matches.iloc[0]

    return {
        "orf_id": orf_id,
        "status": "success",
        "locus_tag": locus_tag,
        "gene_name": gene_record.get("gene_name"),
        "gene_start": int(gene_record["start"]),
        "gene_end": int(gene_record["end"]),
        "strand": gene_record["strand"],
        "protein_id": cds_record["protein_id"],
        "product": cds_record["product"],
    }
```

```python
import requests

def map_refseq_to_uniprot(refseq_protein_id):
    """
    Map a RefSeq protein accession to a UniProt accession using UniProt search.

    Parameters
    ----------
    refseq_protein_id : str
        RefSeq protein accession, for example 'WP_003911169.1'.

    Returns
    -------
    dict
        Mapping result containing status, RefSeq ID, UniProt accession,
        UniProt ID, organism and protein length.
    """

    search_response = requests.get(
        "https://rest.uniprot.org/uniprotkb/search",
        params={
            "query": f"xref:refseq-{refseq_protein_id}",
            "format": "json",
            "size": 10,
        },
        timeout=60,
    )

    if search_response.status_code != 200:
        return {
            "refseq_protein_id": refseq_protein_id,
            "status": "uniprot_search_failed",
            "status_code": search_response.status_code,
            "message": search_response.text[:300],
        }

    search_data = search_response.json()
    results = search_data.get("results", [])

    if len(results) == 0:
        return {
            "refseq_protein_id": refseq_protein_id,
            "status": "no_uniprot_match",
        }

    # Prefer H37Rv / ATCC 25618 if available
    preferred_results = []

    for result in results:
        organism_text = str(result.get("organism", {}))

        if (
            "H37Rv" in organism_text
            or "ATCC 25618" in organism_text
        ):
            preferred_results.append(result)

    if len(preferred_results) > 0:
        selected_result = preferred_results[0]
    else:
        selected_result = results[0]

    return {
        "refseq_protein_id": refseq_protein_id,
        "status": "success",
        "uniprot_accession": selected_result.get("primaryAccession"),
        "uniprot_id": selected_result.get("uniProtkbId"),
        "organism": selected_result.get("organism", {}).get("scientificName"),
        "protein_length": selected_result.get("sequence", {}).get("length"),
        "number_of_matches": len(results),
    }
```
